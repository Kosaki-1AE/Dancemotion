```python
# main.py
# 統合スクリプト（感情・責任・空気感 → 構造知ベクトル出力）

from emotion_vectorizer import EmotionVectorizer
from responsibility_tracker import ResponsibilityTracker
from airflow_analyzer import AirflowAnalyzer
from structure_calculator import StructureCalculator
from structure_visualizer import plot_motion_vector, plot_motion_sequence
from log_manager import LogManager
from recovery_estimator import RecoveryEstimator
from context_modeler import ContextModeler
from gpt_connector import GPTConnector
import numpy as np
import os

# モジュール初期化
emotion_model = EmotionVectorizer()
responsibility = ResponsibilityTracker()
airflow = AirflowAnalyzer()
structure = StructureCalculator()
logger = LogManager()
recovery = RecoveryEstimator()
context = ContextModeler()

# コンテキスト作成・選択
context.create_context("default")
context.switch_context("default")

# GPT連携（APIキーは環境変数 OPENAI_API_KEY に設定）
api_key = os.getenv("OPENAI_API_KEY")
gpt = GPTConnector(api_key) if api_key else None

# 入力例
emotion = "trust"
intensity = 0.8
rhythm = 0.6
stillness = 0.7
tension = 0.2
incoming_arrow = 0.4
outgoing_arrow = 0.5

# 1. 感情ベクトル生成
vec = emotion_model.get_vector(emotion, intensity)
structure.add_action(vec)

# 2. 空気感スコアと責任バランス
airflow.log_state(rhythm, stillness, tension)
airflow_score = airflow.compute_airflow_score()
responsibility.add_incoming(incoming_arrow)
responsibility.add_outgoing(outgoing_arrow)
responsibility_balance = responsibility.current_balance()
recovery.log_airflow(airflow_score)

# 3. フラグで使用要素選択
elements = {
    "use_emotion": True,
    "use_derivative": True,
    "use_airflow": True,
    "use_responsibility": True,
    "use_inertia": True
}

# 4. 構造知ベクトルの出力
predicted_vec = structure.predict_custom_action(vec, airflow_score, responsibility_balance, elements)

# 5. ログ・文脈記録
logger.log(
    step=len(structure.history)-1,
    emotion=emotion,
    intensity=intensity,
    airflow_score=airflow_score,
    responsibility_balance=responsibility_balance,
    predicted_vector=predicted_vec
)
context.log_state(emotion, airflow_score, responsibility_balance, predicted_vec)

# 6. 回復チェック
recovery_info = recovery.detect_recovery()
if recovery_info["detected"]:
    print(f"[RECOVERY DETECTED] Step {recovery_info['index']} → Recovery in {recovery_info['recovery_time']} steps")

# 7. GPT解釈（任意）
if gpt:
    summary = f"Emotion: {emotion}, Airflow: {airflow_score}, Responsibility: {responsibility_balance}"
    try:
        response = gpt.ask(f"Based on the following internal state, what behavior is recommended?\n{summary}")
        print("[GPT Suggestion]:", response)
    except Exception as e:
        print("[GPT Error]:", str(e))

# 8. 可視化
plot_motion_vector(predicted_vec, "Predicted Motion Vector")
plot_motion_sequence(structure.history, "Motion History")

# 9. 出力表示
print("Input vector:", vec)
print("Predicted next vector:", predicted_vec)

# 10. ログ保存
logger.save_csv()
```
